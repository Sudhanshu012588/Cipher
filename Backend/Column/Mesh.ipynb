```julia
import Gmsh: gmsh
```

```julia
using Gmsh

function ColumnMesh(wf, tf, D, L)
    gmsh.initialize()
    gmsh.model.add("Column")
    # Mesh size
    h = 0.25
    p1  = gmsh.model.geo.addPoint(0, 0, 0, h)
    p2  = gmsh.model.geo.addPoint(tf, 0, 0, h)
    p3  = gmsh.model.geo.addPoint(wf, 0, tf, h)

    p4  = gmsh.model.geo.addPoint((wf + tf) / 2, 0, tf, h)
    p5  = gmsh.model.geo.addPoint((wf + tf) / 2, 0, D - tf, h)

    p6  = gmsh.model.geo.addPoint(wf, 0, D - tf, h)
    p7  = gmsh.model.geo.addPoint(wf, 0, D, h)
    p8  = gmsh.model.geo.addPoint(0, 0, D, h)
    p9  = gmsh.model.geo.addPoint(0, 0, D - tf, h)
    p10 = gmsh.model.geo.addPoint((wf - tf) / 2, 0, D - tf, h)
    p11 = gmsh.model.geo.addPoint((wf - tf) / 2, 0, tf, h)
    p12 = gmsh.model.geo.addPoint(0, 0, tf, h)

    l1 = gmsh.model.geo.addLine(p1, p2)
    l2 = gmsh.model.geo.addLine(p2, p3)
    l3 = gmsh.model.geo.addLine(p3, p4)
    l4 = gmsh.model.geo.addLine(p4, p5)
    l5 = gmsh.model.geo.addLine(p5, p6)
    l6 = gmsh.model.geo.addLine(p6, p7)
    l7 = gmsh.model.geo.addLine(p7, p8)
    l8 = gmsh.model.geo.addLine(p8, p9)
    l9 = gmsh.model.geo.addLine(p9, p10)
    l10 = gmsh.model.geo.addLine(p10, p11)
    l11 = gmsh.model.geo.addLine(p11, p12)
    l12 = gmsh.model.geo.addLine(p12, p1)
    curve_loop = gmsh.model.geo.addCurveLoop([
        l1,
        l2,
        l3,
        l4,
        l5,
        l6,
        l7,
        l8,
        l9,
        l10,
        l11,
        l12
    ])

    surface = gmsh.model.geo.addPlaneSurface([
        curve_loop
    ])

    gmsh.model.geo.synchronize()

    gmsh.model.geo.extrude(
        [(2, surface)],
        0,
        L,
        0,
        [20],
        [1.0],
        true
    )
    gmsh.model.geo.synchronize()
    gmsh.model.mesh.generate(3)
    gmsh.write("Column.msh")
    gmsh.finalize()

end
```

```julia
wf = 140.0
tf = 12.4
D  = 300.0
L  = 5000.0
```

```julia
ColumnMesh(wf, tf, D, L)
```

```julia
using Gmsh

function ColumnMeshC(B, t, T, D, L)

    gmsh.initialize()

    gmsh.model.add("C_Column")
    h = 0.25

    p1 = gmsh.model.geo.addPoint(0, 0, 0, h)

    p2 = gmsh.model.geo.addPoint(B, 0, 0, h)

    p3 = gmsh.model.geo.addPoint(B, 0, T, h)

    p4 = gmsh.model.geo.addPoint(t, 0, T, h)

    p5 = gmsh.model.geo.addPoint(t, 0, D - T, h)

    p6 = gmsh.model.geo.addPoint(B, 0, D - T, h)

    p7 = gmsh.model.geo.addPoint(B, 0, D, h)

    p8 = gmsh.model.geo.addPoint(0, 0, D, h)
    l1 = gmsh.model.geo.addLine(p1, p2)

    l2 = gmsh.model.geo.addLine(p2, p3)

    l3 = gmsh.model.geo.addLine(p3, p4)

    l4 = gmsh.model.geo.addLine(p4, p5)

    l5 = gmsh.model.geo.addLine(p5, p6)

    l6 = gmsh.model.geo.addLine(p6, p7)

    l7 = gmsh.model.geo.addLine(p7, p8)

    l8 = gmsh.model.geo.addLine(p8, p1)
    curve_loop = gmsh.model.geo.addCurveLoop([
        l1,
        l2,
        l3,
        l4,
        l5,
        l6,
        l7,
        l8
    ])
    surface = gmsh.model.geo.addPlaneSurface([
        curve_loop
    ])
    gmsh.model.geo.synchronize()
    gmsh.model.geo.extrude(
        [(2, surface)],
        0,
        L,
        0,
        [20],
        [1.0],
        true
    )
    gmsh.model.geo.synchronize()
    gmsh.model.mesh.generate(3)
    gmsh.write("C_Column.msh")
    gmsh.finalize()

end
```

```julia
B = 90.0
t = 7.8
T = 13.6
D = 300.0
L = 5000.0

ColumnMeshC(B, t, T, D, L)
```
